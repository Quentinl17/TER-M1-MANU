```python
import matplotlib.pyplot as plt
import numpy as np

initial = np.loadtxt("SIN_PULSE_initial.txt")
step_1 = np.loadtxt("SIN_PULSE_step_1.txt")
step_2 = np.loadtxt("SIN_PULSE_step_2.txt")
step_3 = np.loadtxt("SIN_PULSE_step_3.txt")
step_4 = np.loadtxt("SIN_PULSE_step_4.txt")
final = np.loadtxt("SIN_PULSE_final.txt")

plt.plot(initial[:,0], initial[:,1], color = 'red', label="Donnée initiale")
plt.plot(step_1[:,0], step_1[:,1], color = 'blue', linewidth=0.5)
plt.plot(step_2[:,0], step_2[:,1], color = 'blue', linewidth=0.5)
plt.plot(step_3[:,0], step_3[:,1], color = 'blue', linewidth=0.5)
plt.plot(step_4[:,0], step_4[:,1], color = 'blue', linewidth=0.5)
plt.plot(final[:,0], final[:,1], color = 'black', label="Solution finale")
plt.xlabel("$x$")
plt.ylabel("$u_0(x)$ et $u_h(x)$")
plt.title("\'Equation de Rayleigh-Stokes 1D")
plt.grid()
plt.legend()
plt.show()
```
