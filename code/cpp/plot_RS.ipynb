```python
import os
import shutil
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
folder = "RS_solutions"

times = np.loadtxt("times_times.txt")

data = []
k = 0
while os.path.exists(f"{folder}/solution_{k}.txt"):
    data.append(np.loadtxt(f"{folder}/solution_{k}.txt"))
    k += 1

print(f"{len(data)} fichiers de solution, {len(times)} instants")

# Échelle commune à toutes les images : min et max globaux + marge
umin = min(d[:, 1].min() for d in data)
umax = max(d[:, 1].max() for d in data)
marge = 0.05 * (umax - umin) if umax > umin else 0.1
ymin, ymax = umin - marge, umax + marge

shutil.rmtree("images", ignore_errors=True)
os.makedirs("images", exist_ok=True)

for k in range(len(data)):
    plt.figure()
    plt.plot(data[0][:, 0], data[0][:, 1], "b--", label="initial")
    plt.plot(data[k][:, 0], data[k][:, 1], "r", label="solution")
    plt.plot(data[-1][:, 0], data[-1][:, 1], "k--", label="final")
    plt.xlabel("$x$")
    plt.ylabel("$u(x,t)$")
    plt.ylim(ymin, ymax)
    plt.title(f"t = {times[k]:.2f}")
    plt.grid()
    plt.legend()
    plt.savefig(f"images/image_{k:04d}.png")
    plt.close()

images = [Image.open(f"images/image_{k:04d}.png") for k in range(len(data))]
images[0].save("RS.gif", save_all=True, append_images=images[1:],
               duration=100, loop=0)
print("GIF créé : RS.gif")
```
